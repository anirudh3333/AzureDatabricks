```python
from pyspark.sql.functions import *
from pyspark.sql.types import * 
from pyspark.sql.window import Window
data = [("101", "2023-12-01", 100), ("101", "2023-12-02", 150), 
        ("102", "2023-12-01", 200), ("102", "2023-12-02", 250)]
columns = ["product_id", "date", "sales"]

df = spark.createDataFrame(data, columns)
df.display()
```

```python
df = df.withColumn('date', col('date').cast(DateType())).withColumn('rowno', row_number().over(Window.partitionBy('product_id').orderBy(col('date').desc()))).filter(col('rowno') ==1).drop('rowno')
display(df)
```
