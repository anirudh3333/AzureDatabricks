```python
# dftest = spark.read.format('parquet').load('abfss://bronze@databrickse2esa.dfs.core.windows.net/orders')
# display(dftest)
dbutils.widgets.text('p_folder_name','orders')
p_folder_name = dbutils.widgets.get('p_folder_name')
print(p_folder_name)
```

## Data reading

```python
df = spark.readStream.format('cloudFiles')\
    .option('cloudFiles.format','parquet')\
        .option('cloudFiles.schemaLocation',f'abfss://bronze@databrickse2esa.dfs.core.windows.net/checkpoint_{p_folder_name}/')\
            .load(f"abfss://source@databrickse2esa.dfs.core.windows.net/{p_folder_name}")
```

## Data writing

```python
df.writeStream.format('parquet')\
    .option('checkpointLocation',f'abfss://bronze@databrickse2esa.dfs.core.windows.net/checkpoint_{p_folder_name}/')\
        .trigger(once=True)\
            .start(f'abfss://bronze@databrickse2esa.dfs.core.windows.net/{p_folder_name}/')
```
