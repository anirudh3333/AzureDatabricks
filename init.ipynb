```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import Window
```

```python
df = spark.read.format('csv').option('inferSchema', True).option('header', True).load('/Volumes/databricks_cata/bronze/myvol/fact_bookings.csv')
display(df)
```

```python
df.groupBy('booking_date').agg(round(sum('amount'),2)).orderBy(col('booking_date').asc()).display()
```
